```python
prices = list(range(1, 1_000_001))

start = time.time()
discounted_loop = [p * 0.95 for p in prices]
loop_time = time.time() - start
print(f"Loop version: {loop_time :.4f} seconds")

prices_np = np.arange(1, 1_000_001)

start = time.time()
discounted_np = prices_np * 0.95
np_time = time.time() - start
print(f"NumPy version: { np_time :.4f} seconds")

print(f"NumPy was about {loop_time / np_time:.0f}x faster")
```

```python
a = np.array([10, 20, 30, 40])
zeros = np.zeros((2, 3))
ones = np.ones((3, 3))
seq = np.arange(0, 10, 2)
lin = np.linspace(0, 1, 5)

print("a:", a)
print("zeros:\n", zeros)
print("ones:\n", ones)
print("seq:", seq)
print("lin:", lin)
```

```python
matrix = np.array([[1,2,3],[4,5,6]])

print("shape:", matrix.shape)
print("ndim:", matrix.ndim)
print("size:" ,matrix.size)
print("dtype:", matrix.dtype)
```

```python
flat = np.arange(12)
print("flat:", flat)

reshaped = flat.reshape(3,4)
print("reshaped (3x4):\n", reshaped)

back_to_flat = reshaped.flatten()
print("flattened_again:", back_to_flat)
```

```python
sales = np.array([120, 85, 200, 0, 60, 300, 45])

print("first element:", sales[0])
print("last element:", sales[-1])
print("first 3:", sales[:3])
print("every other:", sales[::2])

mask = sales > 100
print("mask:", mask)
print("sales above 100:", sales[mask])
print("days with zero sales:", sales[sales == 0])
print("between 50 and 150:", sales[(sales > 50) & (sales < 150)])
```

```python
units_sold = np.array([10, 25, 8, 40])
unit_price = np.array([15000, 22000, 9500, 30000])

revenue = units_sold * unit_price
print("revenue per product:", revenue)
print("total revenue:", revenue.sum())
```

```python
store_sales = np.array([
    [100, 200, 150, 300],
    [120, 180, 170, 250],
    [90, 210, 160, 280]
])

discount = 0.10
print("after 10% discount:\n", store_sales * (1 - discount))

product_bonus = np.array([5, 0, 10, 2])
print("\nwith per-product bonus:\n", store_sales + product_bonus)
```

```python
print("total per store (axis=1, across columns):", store_sales.sum(axis=1))
print("total per product (axis=0, down rows):", store_sales.sum(axis=0))
print("overall mean:", store_sales.mean())
print("best-selling product per store:", store_sales.argmax(axis=1))
```

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6], [7,8]])

print("A:\n", A)
print("A transposed:\n", A.T)
print("\nElement-wise multiply(A * B):\n", A * B)
print("\nMatrix multiply (A @ B):\n", A @ B)
```

```python
holdings = np.array([
    [10, 5, 0, 20],
    [0, 15, 10, 5],
    [8, 8, 8, 5]
])

prices = np.array([50000, 120000, 75000, 30000])

portfolio_value = holdings @ prices
print("value of each portfolio (KRW):", portfolio_value)
```

```python
I = np.eye(3)
print("identity matrix:\n", I)

M = np.array([[2,1], [1,3]])
M_inv = np.linalg.inv(M)
print("\nM inverse\n:", M_inv)
print("M @ M_inv (should be ~identity):\n", np.round(M @ M_inv, 2))

coeffs = np.array([[2,1], [1,3]])
results = np.array([5, 10])
solution =np.linalg.solve(coeffs, results)
print("\nsolution [x,y]:", solution)
```

```python
np.random.seed(42)

uniform_samples = np.random.rand(5)
integers = np.random.randint(1, 7, size=10)
normal_samples = np.random.normal(loc=170, scale=8, size=5)

print("uniform:", uniform_samples)
print("dice rolls:", integers)
print("normal samples:", normal_samples)
```

```python
np.random.seed(42)
rolls = np.random.randint(1, 7, size=100_000)

prob_six = (rolls == 6).mean()
print(f"P(rolling a 6): {prob_six:.4f} (theoretical: {1/6:.4f})")

avg_roll = rolls.mean()
print(f"average roll: {avg_roll:.3f} (theoretical: 3.5)")
```

```python
np.random.seed(42)
days = 30
daily_returns = np.random.normal(loc=0.001, scale=0.02, size=days)

price = 10000 * np.cumprod(1 * daily_returns)
print("simulated closing prices:\n", np.round(price, 0))
print("\nfinal price after", days, "days:", round(price[-1], 0))
```
